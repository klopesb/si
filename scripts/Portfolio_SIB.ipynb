### Exercise 10: Implementing the StackingClassifier ensemble

10.1) In the "ensemble" sub-package add the "stacking_classifier.py" module. Implement
the StackingClassifier class by taking into consideration the structure of the class shown
in the next slides.

```python
import numpy as np
from si.base.model import Model
from si.data.dataset import Dataset
from si.metrics.accuracy import accuracy


class StackingClassifier(Model):
    """
    Stacking Classifier ensemble model.
    
    The StackingClassifier uses an ensemble of base models to generate predictions.
    These predictions are then used as features to train a final model that makes
    the ultimate predictions.
    
    Parameters
    ----------
    models : list
        List of base models to use in the ensemble
    final_model : Model
        The model to make the final predictions based on base model outputs
        
    Attributes
    ----------
    models : list
        The fitted base models
    final_model : Model
        The fitted final model
    """
    
    def __init__(self, models: list, final_model: Model, **kwargs):
        """
        Initialize the StackingClassifier.
        
        Parameters
        ----------
        models : list
            Initial set of base models
        final_model : Model
            The model to make the final predictions
        """
        super().__init__(**kwargs)
        self.models = models
        self.final_model = final_model
        
    def _fit(self, dataset: Dataset) -> 'StackingClassifier':
        """
        Train the stacking classifier.
        
        Algorithm:
        1. Train the initial set of models
        2. Get predictions from the initial set of models
        3. Train the final model with the predictions of the initial set of models
        4. Return itself (self)
        
        Parameters
        ----------
        dataset : Dataset
            The dataset to fit the model to
            
        Returns
        -------
        self : StackingClassifier
            The fitted model
        """
        # Step 1: Train the initial set of models
        for model in self.models:
            model.fit(dataset)
        
        # Step 2: Get predictions from the initial set of models
        base_predictions = np.column_stack([
            model.predict(dataset) for model in self.models
        ])
        
        # Step 3: Train the final model with the predictions of the initial set of models
        # Create a new dataset with base predictions as features and original labels
        stacked_dataset = Dataset(X=base_predictions, y=dataset.y)
        self.final_model.fit(stacked_dataset)
        
        # Step 4: Return itself
        return self
    
    def _predict(self, dataset: Dataset) -> np.ndarray:
        """
        Predict class labels using the stacking classifier.
        
        Algorithm:
        1. Get predictions from the initial set of models
        2. Get the final predictions using the final model and the 
           predictions of the initial set of models
        
        Parameters
        ----------
        dataset : Dataset
            The dataset to predict the classes of
            
        Returns
        -------
        predictions : np.ndarray
            Predicted class labels
        """
        # Step 1: Get predictions from the initial set of models
        base_predictions = np.column_stack([
            model.predict(dataset) for model in self.models
        ])
        
        # Step 2: Get the final predictions using the final model
        # Create a new dataset with base predictions as features
        stacked_dataset = Dataset(X=base_predictions, y=dataset.y)
        final_predictions = self.final_model.predict(stacked_dataset)
        
        return final_predictions
    
    def _score(self, dataset: Dataset, predictions: np.ndarray) -> float:
        """
        Compute the accuracy score of the stacking classifier.
        
        Algorithm:
        1. Compute the accuracy between predicted and real values
        
        Parameters
        ----------
        dataset : Dataset
            The dataset to evaluate the model on
        predictions : np.ndarray
            An array with the predictions
            
        Returns
        -------
        accuracy : float
            The accuracy score (proportion of correct predictions)
        """
        # Compute the accuracy between predicted and real values
        return accuracy(dataset.y, predictions)
```

### Exercise 11: Implementing the randomized_search_cv function.

11.1) In the "model_selection" sub-package add the "randomized_search.py"
module. Implement the randomized_search_cv function by considering the
structure presented in the next slides.

```python
from typing import Dict, Callable, List, Any
import numpy as np
import itertools
from si.base.model import Model
from si.data.dataset import Dataset



def randomized_search_cv(model: Model,
                         dataset: Dataset,
                         hyperparameter_grid: Dict[str, List[Any]],
                         scoring: Callable = None,
                         cv: int = 3,
                         n_iter: int = 10,
                         seed: int = None) -> Dict[str, Any]:
    """
    Perform randomized search cross-validation on the given model and dataset.
    
    This function randomly samples n_iter combinations from the hyperparameter grid,
    evaluates each combination using k-fold cross-validation, and returns the best
    performing hyperparameters.

    Parameters
    ----------
    model : Model
        The model to validate
    dataset : Dataset
        The validation dataset
    hyperparameter_grid : Dict[str, List[Any]]
        Dictionary with the hyperparameter names as keys and lists of possible
        values to search as values
    scoring : Callable, optional
        Score function to use. If None, the model's score method will be used
    cv : int, default=3
        Number of cross-validation folds
    n_iter : int, default=10
        Number of random hyperparameter combinations to test
    seed : int, optional
        Random seed for reproducibility

    Returns
    -------
    results : Dict[str, Any]
        Dictionary with the following keys:
        - 'hyperparameters': list of all hyperparameter combinations tested
        - 'scores': list of mean scores obtained for each combination
        - 'best_hyperparameters': best combination of hyperparameters
        - 'best_score': best mean score achieved

    Raises
    ------
    ValueError
        If any hyperparameter in the grid does not exist in the model

    """

    # Step 1: Check if the provided hyperparameters are valid
    for hyperparameter in hyperparameter_grid.keys():
        if not hasattr(model, hyperparameter):
            raise ValueError(f"Model {model.__class__.__name__} does not have "
                           f"hyperparameter '{hyperparameter}'")
    
    # Step 2: Get n_iter hyperparameter combinations
    # Generate all possible combinations
    hyperparameter_names = list(hyperparameter_grid.keys())
    hyperparameter_values = list(hyperparameter_grid.values())
    all_combinations = list(itertools.product(*hyperparameter_values))
    
    # Randomly sample n_iter combinations (or all if n_iter > total combinations)
    if seed is not None:
        np.random.seed(seed)
    
    n_combinations = min(n_iter, len(all_combinations))
    random_indices = np.random.choice(len(all_combinations), 
                                     size=n_combinations, 
                                     replace=False)
    selected_combinations = [all_combinations[i] for i in random_indices]
    
    # Initialize results storage
    all_hyperparameters = []
    all_scores = []
    best_score = -np.inf
    best_hyperparameters = None
    
    # Step 3-6: Iterate through all selected hyperparameter combinations
    for combination in selected_combinations:
        # Step 3: Set the model hyperparameters with the current combination
        current_hyperparameters = {}
        for hyperparameter_name, hyperparameter_value in zip(hyperparameter_names, combination):
            setattr(model, hyperparameter_name, hyperparameter_value)
            current_hyperparameters[hyperparameter_name] = hyperparameter_value
        
        # Step 4: Cross validate the model using k_fold_cross_validation
        fold_scores = k_fold_cross_validation(model, dataset, scoring=scoring, cv=cv, seed=seed)
        
        # Step 5: Save the mean of the scores and respective hyperparameters
        mean_score = np.mean(fold_scores)
        all_hyperparameters.append(current_hyperparameters.copy())
        all_scores.append(mean_score)
        
        # Step 7: Update best score and hyperparameters if current is better
        if mean_score > best_score:
            best_score = mean_score
            best_hyperparameters = current_hyperparameters.copy()
    
    # Step 8: Return results dictionary
    results = {
        'hyperparameters': all_hyperparameters,
        'scores': all_scores,
        'best_hyperparameters': best_hyperparameters,
        'best_score': best_score
    }
    
    return results

```

• Build, train and evaluate a NN based on the following instructions:
- The training dataset has 32 features
- The task is binary classification
- Use the SGD optimizer
- Use the BinaryCrossEntropy loss
- Use the accuracy metric
- The model should contain:
• Dense layer 1
• ReLU activation layer 1
• Dense layer 2
• ReLU activation layer 2
• Output Dense layer
• Sigmoid activation layer
- The dense layers should reduce the number of units to half (except the last one)
- Train the NN for 100 epochs, with batch size of 16 with a learning rate of 0.01.
- Test the model with k fold cross validation (hint: use the functions implemented in
class 8)

```python
import numpy as np
import pandas as pd
from si.data.dataset import Dataset
from si.neural_networks.neural_network import NeuralNetwork
from si.neural_networks.layers import DenseLayer
from si.neural_networks.activation import ReLUActivation, SigmoidActivation
from si.neural_networks.optimizers import SGD
from si.neural_networks.losses import BinaryCrossEntropy
from si.model_selection.cross_validate import k_fold_cross_validation
from si.metrics.accuracy import accuracy


def build_neural_network(n_features: int = 32, learning_rate: float = 0.01) -> NeuralNetwork:
    """
    Build a neural network for binary classification.
    
    Architecture:
    - Input: 32 features
    - Dense layer 1: 16 units (half of input)
    - ReLU activation 1
    - Dense layer 2: 8 units (half of previous)
    - ReLU activation 2
    - Output Dense layer: 1 unit (binary classification)
    - Sigmoid activation (output probabilities)
    
    Parameters
    ----------
    n_features : int
        Number of input features (default: 32)
    learning_rate : float
        Learning rate for SGD optimizer (default: 0.01)
    
    Returns
    -------
    NeuralNetwork
        The configured neural network
    """
    # Initialize the neural network
    model = NeuralNetwork(
        epochs=100,
        batch_size=16,
        optimizer=SGD,
        learning_rate=learning_rate,
        loss=BinaryCrossEntropy,
        metric=accuracy,
        verbose=True
    )
    
    # Add layers to the network
    # Dense layer 1: Reduce to half (32 -> 16)
    model.add(DenseLayer(n_units=n_features // 2, input_shape=(n_features,)))
    model.add(ReLUActivation())
    
    # Dense layer 2: Reduce to half again (16 -> 8)
    model.add(DenseLayer(n_units=n_features // 4))
    model.add(ReLUActivation())
    
    # Output layer: Single unit for binary classification
    model.add(DenseLayer(n_units=1))
    model.add(SigmoidActivation())
    
    return model


def train_and_evaluate(dataset: Dataset, n_features: int = 32, 
                      learning_rate: float = 0.01, cv: int = 5) -> dict:
    """
    Train and evaluate the neural network using k-fold cross-validation.
    
    Parameters
    ----------
    dataset : Dataset
        The dataset to train and evaluate on
    n_features : int
        Number of input features
    learning_rate : float
        Learning rate for SGD optimizer
    cv : int
        Number of folds for cross-validation
    
    Returns
    -------
    dict
        Dictionary containing evaluation results
    """
    print("=" * 70)
    print("Neural Network Configuration")
    print("=" * 70)
    print(f"Input features: {n_features}")
    print(f"Architecture: {n_features} -> {n_features//2} -> {n_features//4} -> 1")
    print(f"Optimizer: SGD (learning_rate={learning_rate})")
    print(f"Loss: BinaryCrossEntropy")
    print(f"Metric: Accuracy")
    print(f"Epochs: 100")
    print(f"Batch size: 16")
    print(f"Cross-validation folds: {cv}")
    print("=" * 70)
    print()
    
    # Build the model
    model = build_neural_network(n_features=n_features, learning_rate=learning_rate)
    
    # Display model summary
    print("Model Architecture:")
    print("-" * 70)
    print(f"{'Layer':<5} {'Type':<25} {'Output Shape':<20} {'Parameters':<15}")
    print("-" * 70)
    total_params = 0
    for i, layer in enumerate(model.layers, 1):
        layer_params = layer.parameters()
        total_params += layer_params
        print(f"{i:<5} {layer.layer_name():<25} {str(layer.output_shape()):<20} {layer_params:<15}")
    print("-" * 70)
    print(f"Total parameters: {total_params:,}")
    print("=" * 70)
    print()
    
    # Perform k-fold cross-validation
    print("Starting K-Fold Cross-Validation...")
    print("=" * 70)
    
    scores = k_fold_cross_validation(
        model=model,
        dataset=dataset,
        scoring=accuracy,
        cv=cv,
        seed=42
    )
    
    # Calculate statistics
    results = {
        'scores': scores,
        'mean_score': np.mean(scores),
        'std_score': np.std(scores),
        'min_score': np.min(scores),
        'max_score': np.max(scores)
    }
    
    return results


def display_results(results: dict):
    """
    Display the cross-validation results.
    
    Parameters
    ----------
    results : dict
        Dictionary containing evaluation results
    """
    print("\n" + "=" * 70)
    print("Cross-Validation Results")
    print("=" * 70)
    print(f"Individual fold scores:")
    for i, score in enumerate(results['scores'], 1):
        print(f"  Fold {i}: {score:.4f} ({score*100:.2f}%)")
    print("-" * 70)
    print(f"Mean Accuracy: {results['mean_score']:.4f} ({results['mean_score']*100:.2f}%)")
    print(f"Std Deviation: {results['std_score']:.4f}")
    print(f"Min Accuracy:  {results['min_score']:.4f} ({results['min_score']*100:.2f}%)")
    print(f"Max Accuracy:  {results['max_score']:.4f} ({results['max_score']*100:.2f}%)")
    print("=" * 70)


# Main execution
if __name__ == "__main__":
    print("\n" + "=" * 70)
    print("Exercise 16: Neural Network Binary Classification")
    print("=" * 70)
    print()
    
    # Option 1: Load from CSV (adjust path as needed)
    try:
        print("Loading dataset from CSV...")
        df = pd.read_csv('../datasets/breast_bin/breast-bin.csv', header=None)
        X = df.iloc[:, :-1].to_numpy()
        y = df.iloc[:, -1].to_numpy()
        
        # Ensure y is reshaped for binary classification
        if len(y.shape) == 1:
            y = y.reshape(-1, 1)
        
        n_features = X.shape[1]
        dataset = Dataset(X, y)
        
        print(f"✓ Dataset loaded successfully!")
        print(f"  - Samples: {X.shape[0]}")
        print(f"  - Features: {n_features}")
        print(f"  - Target shape: {y.shape}")
        print(f"  - Target unique values: {np.unique(y)}")
        print()
        
    except FileNotFoundError:
        print("Dataset file not found. Creating synthetic dataset...")
        # Option 2: Create synthetic dataset with 32 features
        n_samples = 1000
        n_features = 32
        
        np.random.seed(42)
        X = np.random.randn(n_samples, n_features)
        # Binary classification based on sum of first 10 features
        y = (np.sum(X[:, :10], axis=1) > 0).astype(float).reshape(-1, 1)
        
        dataset = Dataset(X, y)
        
        print(f"✓ Synthetic dataset created!")
        print(f"  - Samples: {n_samples}")
        print(f"  - Features: {n_features}")
        print(f"  - Task: Binary classification")
        print(f"  - Class distribution: {np.unique(y, return_counts=True)}")
        print()
    
    # Train and evaluate the model with k-fold cross-validation
    results = train_and_evaluate(
        dataset=dataset,
        n_features=n_features,
        learning_rate=0.01,
        cv=5
    )
    
    # Display results
    display_results(results)
    
    print("\n" + "=" * 70)
    print("Training a final model on the full dataset...")
    print("=" * 70)
    print()
    
    # Train a final model on the full dataset
    final_model = build_neural_network(n_features=n_features, learning_rate=0.01)
    final_model.fit(dataset)
    
    # Make predictions
    predictions = final_model.predict(dataset)
    
    # Binarize predictions (threshold at 0.5)
    predictions_binary = (predictions >= 0.5).astype(float)
    
    # Calculate final accuracy
    final_accuracy = accuracy(dataset.y, predictions_binary)
    
    print(f"\nFinal model accuracy on full dataset: {final_accuracy:.4f} ({final_accuracy*100:.2f}%)")
    
    # Show some example predictions
    print("\nExample predictions (first 10 samples):")
    print("-" * 70)
    print(f"{'Sample':<10} {'True Label':<15} {'Predicted Prob':<20} {'Predicted Label':<20}")
    print("-" * 70)
    for i in range(min(10, len(predictions))):
        print(f"{i+1:<10} {dataset.y[i][0]:<15.0f} {predictions[i][0]:<20.4f} {predictions_binary[i][0]:<20.0f}")
    print("-" * 70)
    
    print("\n" + "=" * 70)
    print("Exercise 16 Complete!")
    print("=" * 70)
    
    # Summary
    print("\nSummary:")
    print(f"✓ Built neural network with architecture: {n_features} -> {n_features//2} -> {n_features//4} -> 1")
    print(f"✓ Layers: DenseLayer -> ReLU -> DenseLayer -> ReLU -> DenseLayer -> Sigmoid")
    print(f"✓ Used SGD optimizer with learning rate 0.01")
    print(f"✓ Used BinaryCrossEntropy loss")
    print(f"✓ Used Accuracy metric")
    print(f"✓ Trained for 100 epochs with batch size 16")
    print(f"✓ Evaluated with {len(results['scores'])}-fold cross-validation")
    print(f"✓ Mean CV Accuracy: {results['mean_score']:.4f} ± {results['std_score']:.4f}")
    print(f"✓ Final model accuracy: {final_accuracy:.4f}")
    print("=" * 70)
```
